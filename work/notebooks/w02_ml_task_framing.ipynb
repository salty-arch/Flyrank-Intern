## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

My taget priority is, calculating the Gap which is equal to median CTR minus the page CTR and then multiplying that Gap with the page impressions so if a page has huge Gap and high Impressions then its score can ultimately be higher.

This target is a proxy I've defined/constructed from a formula and with it I measured the scoring, not an outcome directly observed in the data. This score is decision-support tool and it helps the SEO person to prioritize which pages to review.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A simple rule like 'flag if CTR is below X' fails here because it judges each page against the same bar, however CTR changes depending on the position tier. For example a deep tier page with CTR of 0.1 might be normal but that same CTR for a high tier page is problematic. ML here beats fixed rules like these which only go by one rule when there are several other aspects to look and consider too.
